# **Import necessary libraries and dataset**

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('IBM Telco Customer Churn.csv')
```

```python
# Check basic info and preview
print(f"Dataset Shape: {df.shape}")
df.info()
df.head()
```

# **Clean and Transform the Data**

```python
# 1. Standardize column names to lowercase snake_case
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

# 2. Fix TotalCharges data type (convert whitespace to 0 and cast to float)
df['totalcharges'] = df['totalcharges'].replace(' ', '0')
df['totalcharges'] = pd.to_numeric(df['totalcharges'], errors='coerce').fillna(0)

# 3. Create a numeric churn flag (1 for Yes, 0 for No) for easy KPI calculations
df['churn_binary'] = df['churn'].apply(lambda x: 1 if x.lower() == 'yes' else 0)

# 4. Create Tenure Cohort Buckets
bins = [0, 12, 24, 48, 72]
labels = ['0-1 Year', '1-2 Years', '2-4 Years', '4+ Years']
df['tenure_cohort'] = pd.cut(df['tenure'], bins=bins, labels=labels, right=True, include_lowest=True)

# Preview the cleaned data
print("Data cleaned successfully!")
df[['customerid', 'tenure', 'tenure_cohort', 'monthlycharges', 'totalcharges', 'churn_binary']].head()
```

# **Download cleaned data to load into BigQuery**

```python
# Save the cleaned DataFrame to a CSV
df.to_csv('cleaned_telco_churn.csv', index=False)

# Trigger the download to your local machine
from google.colab import files
files.download('cleaned_telco_churn.csv')
```
